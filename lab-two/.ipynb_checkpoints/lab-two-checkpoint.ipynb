### 1. Setup and dataset path
Import the required libraries and configure input/output directories. Place the supplied CSV files inside data/ beside this notebook.

```python
# Uncomment if needed:
# %pip install pandas numpy matplotlib seaborn scikit-learn jupyter

from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
DATA_DIR = Path("data")
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Could not find {DATA_DIR.resolve()}. Create a 'data' folder beside this notebook "
        "and place the supplied CSV files in it."
    )

print("Notebook directory:", Path.cwd())
print("Dataset directory:", DATA_DIR.resolve())
print("CSV files found:", sorted(p.name for p in DATA_DIR.glob("*.csv")))
```

## 2. Inspect available datasets
Check which CSV files are present and review their dimensions before selecting regression tasks.

```python
def load_csv(filename, **kwargs):
    path = DATA_DIR / filename
    if not path.exists():
        raise FileNotFoundError(f"Required dataset not found: {path.resolve()}")
    return pd.read_csv(path, **kwargs)

available_files = sorted(DATA_DIR.glob("*.csv"))
inventory = []
for path in available_files:
    try:
        sample = pd.read_csv(path, nrows=3)
        inventory.append({
            "file": path.name,
            "columns_in_sample": len(sample.columns),
            "sample_rows_read": len(sample),
            "size_kb": round(path.stat().st_size / 1024, 1)
        })
    except Exception as exc:
        inventory.append({"file": path.name, "error": str(exc)})

display(pd.DataFrame(inventory))
```

### 3. Clean and prepare datasets
Four continuous targets are used for regression: concrete strength, insurance charges, steel energy usage, and retail quantity. The other files are retained in data/, but datasets whose main outcomes are categorical are better suited to classification.

```python
# 3A. Concrete: first row in some copies may be a units/metadata row.
concrete_raw = load_csv("concrete-data.csv")
concrete = concrete_raw.copy()
concrete_target = "Concrete compressive strength(MPa, megapascals) "
concrete[concrete_target] = pd.to_numeric(concrete[concrete_target], errors="coerce")
concrete = concrete.dropna(subset=[concrete_target]).copy()
for col in concrete.columns:
    concrete[col] = pd.to_numeric(concrete[col], errors="coerce")
concrete = concrete.dropna(subset=[concrete_target])
concrete = concrete.loc[concrete[concrete_target] > 0].reset_index(drop=True)
concrete_task = {
    "name": "Concrete strength",
    "df": concrete,
    "target": concrete_target,
    "features": [c for c in concrete.columns if c != concrete_target],
    "sample_size": None
}

# 3B. Insurance: predict charges from demographic and policy-related features.
insurance = load_csv("insurance.csv")
insurance["charges"] = pd.to_numeric(insurance["charges"], errors="coerce")
insurance = insurance.dropna(subset=["charges"]).copy()
insurance_task = {
    "name": "Medical insurance charges",
    "df": insurance,
    "target": "charges",
    "features": [c for c in insurance.columns if c != "charges"],
    "sample_size": None
}

# 3C. Steel industry: parse the date and derive time features; predict energy use.
steel = load_csv("steel-industry.csv")
steel["Usage_kWh"] = pd.to_numeric(steel["Usage_kWh"], errors="coerce")
steel["date"] = pd.to_datetime(steel["date"], errors="coerce", dayfirst=True)
steel = steel.dropna(subset=["Usage_kWh"]).copy()
if "date" in steel.columns:
    steel["hour"] = steel["date"].dt.hour
    steel["month"] = steel["date"].dt.month
    steel["day"] = steel["date"].dt.day
    steel["dayofweek_num"] = steel["date"].dt.dayofweek
    steel = steel.drop(columns=["date"])
steel_task = {
    "name": "Steel industry energy usage",
    "df": steel,
    "target": "Usage_kWh",
    "features": [c for c in steel.columns if c != "Usage_kWh"],
    "sample_size": None
}

tasks = [concrete_task, insurance_task, steel_task]

summary = []
for task in tasks:
    df = task["df"]
    summary.append({
        "dataset/task": task["name"],
        "rows after cleaning": len(df),
        "predictor count": len(task["features"]),
        "target": task["target"],
        "missing cells": int(df[task["features"] + [task["target"]]].isna().sum().sum()),
        "duplicate rows": int(df.duplicated().sum())
    })
display(pd.DataFrame(summary))
```

### 4. Exploratory data analysis
Review descriptive statistics, target distributions, and numeric correlations. These checks help identify skew, unusual values, and potentially useful predictors before model fitting.

```python
for task in tasks:
    df, target = task["df"], task["target"]
    print("\n" + "=" * 90)
    print(task["name"], "| target:", target)
    display(df[task["features"] + [target]].describe(include="all").T.head(20))
    print("Target summary:")
    display(df[target].describe().to_frame().T)
    fig, ax = plt.subplots()
    sns.histplot(x=pd.to_numeric(df[target], errors="coerce").dropna(), kde=False, ax=ax)
    ax.set_title(f"{task['name']}: target distribution")
    ax.set_xlabel(target)
    plt.tight_layout()
    plt.show()

    numeric_df = df.select_dtypes(include=np.number)
    if target in numeric_df.columns and numeric_df.shape[1] > 1:
        corr = numeric_df.corr(numeric_only=True)[target].drop(target).sort_values()
        corr = pd.concat([corr.head(5), corr.tail(5)]).drop_duplicates()
        fig, ax = plt.subplots(figsize=(8, max(3, len(corr) * 0.35)))
        corr.sort_values().plot(kind="barh", ax=ax)
        ax.set_title(f"{task['name']}: numeric feature correlation with target")
        ax.set_xlabel("Pearson correlation")
        plt.tight_layout()
        plt.show()
```

### 5. Regression methods used
- Multiple Linear Regression: predicts one continuous target using multiple input features.
- Polynomial Regression (degree 2): adds a squared term to model a curved relationship; for a clear baseline, it uses the strongest numeric predictor.
- Ridge Regression (L2): linear regression with coefficient shrinkage to help control overfitting.
- Lasso Regression (L1): linear regression with coefficient shrinkage that can reduce some coefficients to zero.
Simple Linear Regression is represented by the single-predictor setup used for the polynomial baseline's underlying relationship; the comparison focuses on multiple linear regression, polynomial regression, Ridge, and Lasso to keep the model comparison manageable. Logistic regression is excluded because it predicts categorical outcomes rather than a continuous dependent variable.

```python
def make_preprocessor(X, scale_numeric=True):
    """Impute missing values, scale numeric columns, and encode categories."""
    numeric_features = X.select_dtypes(include=np.number).columns.tolist()
    categorical_features = [c for c in X.columns if c not in numeric_features]

    numeric_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))
    numeric_pipe = Pipeline(numeric_steps)

    categorical_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore", min_frequency=2))
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipe, numeric_features),
        ("categorical", categorical_pipe, categorical_features)
    ], remainder="drop")


def strongest_numeric_feature(X, y):
    """Choose a numeric feature with the strongest absolute linear correlation to y."""
    numeric = X.select_dtypes(include=np.number)
    if numeric.shape[1] == 0:
        # A fallback for datasets with only categorical predictors.
        return X.columns[0]
    correlations = numeric.apply(
        lambda col: pd.to_numeric(col, errors="coerce").corr(y)
    ).abs().dropna()
    return correlations.idxmax() if not correlations.empty else numeric.columns[0]


def regression_model_specs():
    """Return the requested regression methods; logistic regression is excluded."""
    return {
        "Multiple Linear Regression": ("multiple", LinearRegression()),
        "Polynomial Regression (degree 2)": ("polynomial", LinearRegression()),
        "Ridge Regression": ("multiple", Ridge(alpha=1.0)),
        "Lasso Regression": ("multiple", Lasso(alpha=0.01, max_iter=10000))
    }
```

### 6. Train models and evaluate performance

All methods use the same train/test split within each dataset. Compare MAE, MSE, RMSE, and R²; lower error values are better, while higher R² is better. The best model is selected using test RMSE.

```python
results = []
fitted_models = {}
predictions = {}
splits = {}
selected_features = {}

for task in tasks:
    name, df, target = task["name"], task["df"], task["target"]
    X_all = df[task["features"]].copy()
    y = pd.to_numeric(df[target], errors="coerce")
    valid = y.notna()
    X_all, y = X_all.loc[valid], y.loc[valid]

    # Keep a common holdout split so all regression methods are compared fairly.
    X_train_all, X_test_all, y_train, y_test = train_test_split(
        X_all, y, test_size=0.2, random_state=RANDOM_STATE
    )
    splits[name] = (X_train_all, X_test_all, y_train, y_test)

    # Simple and polynomial regression use one numeric predictor for an interpretable baseline.
    best_feature = strongest_numeric_feature(X_train_all, y_train)
    selected_features[name] = best_feature

    fitted_models[name] = {}
    predictions[name] = {}

    for model_name, (model_type, estimator) in regression_model_specs().items():
        if model_type == "polynomial":
            # Polynomial terms are created from one numeric feature to avoid a huge
            # feature explosion on categorical/high-cardinality datasets.
            feature = best_feature
            X_train = X_train_all[[feature]].copy()
            X_test = X_test_all[[feature]].copy()
            preprocessor = Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler()),
                ("polynomial", PolynomialFeatures(degree=2, include_bias=False))
            ])
        else:
            X_train, X_test = X_train_all, X_test_all
            preprocessor = make_preprocessor(X_train)

        pipe = Pipeline([("preprocess", preprocessor), ("model", estimator)])
        pipe.fit(X_train, y_train)
        y_pred = pipe.predict(X_test)

        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        rmse = float(np.sqrt(mse))
        r2 = r2_score(y_test, y_pred)

        results.append({
            "Dataset": name,
            "Model": model_name,
            "MAE": mae,
            "MSE": mse,
            "RMSE": rmse,
            "R²": r2,
            "Predictor setup": (
                f"Single feature: {best_feature}" if model_type == "polynomial"
                else "All selected features"
            ),
            "Train rows": len(X_train),
            "Test rows": len(X_test)
        })
        fitted_models[name][model_name] = pipe
        predictions[name][model_name] = (y_test, y_pred)

results_df = pd.DataFrame(results).sort_values(["Dataset", "RMSE"]).reset_index(drop=True)
display(results_df.round(4))
results_df.to_csv(OUTPUT_DIR / "regression_model_metrics.csv", index=False)
print("Saved:", (OUTPUT_DIR / "regression_model_metrics.csv").resolve())
```

### 7. Compare models with bar charts
Each chart compares the candidate regression methods within a dataset. Error scales differ across datasets, so comparisons should primarily be made within each dataset.

```python
metric_order = ["MAE", "MSE", "RMSE", "R²"]
for metric in metric_order:
    g = sns.catplot(
        data=results_df, x="Model", y=metric, col="Dataset",
        kind="bar", col_wrap=2, sharey=False, height=4, aspect=1.25
    )
    g.set_xticklabels(rotation=25, ha="right")
    g.fig.suptitle(f"Model comparison by {metric}", y=1.03)
    g.tight_layout()
    plt.show()
```

### 8. Actual vs. predicted and residual plots
Predictions close to the diagonal indicate a better fit. Residuals should be scattered around zero without obvious patterns.

```python
for task in tasks:
    name = task["name"]
    # Plot the best model by test RMSE, then compare residual behavior.
    subset = results_df[results_df["Dataset"] == name]
    best_model_name = subset.sort_values("RMSE").iloc[0]["Model"]
    y_test, y_pred = predictions[name][best_model_name]
    residuals = np.asarray(y_test) - np.asarray(y_pred)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
    axes[0].scatter(y_test, y_pred, alpha=0.55, edgecolor="none")
    lo = min(float(np.min(y_test)), float(np.min(y_pred)))
    hi = max(float(np.max(y_test)), float(np.max(y_pred)))
    axes[0].plot([lo, hi], [lo, hi], linestyle="--")
    axes[0].set_xlabel("Actual")
    axes[0].set_ylabel("Predicted")
    axes[0].set_title(f"{name}\nActual vs. predicted — {best_model_name}")

    axes[1].scatter(y_pred, residuals, alpha=0.55, edgecolor="none")
    axes[1].axhline(0, linestyle="--")
    axes[1].set_xlabel("Predicted")
    axes[1].set_ylabel("Residual (actual − predicted)")
    axes[1].set_title("Residual plot")
    plt.tight_layout()
    plt.show()
```

### 9. Visualize a feature-target relationship
A regression plot illustrates the relationship between the strongest numeric predictor and the target. This is a univariate visual aid, not the full multivariable model.

```python
for task in tasks:
    df, target, name = task["df"], task["target"], task["name"]
    numeric = df.select_dtypes(include=np.number)
    candidates = [c for c in numeric.columns if c != target]
    if not candidates:
        print(f"No numeric predictor available for {name}; skipping regression plot.")
        continue
    corr = numeric[candidates + [target]].corr(numeric_only=True)[target].drop(target).abs()
    feature = corr.idxmax()
    plot_df = df[[feature, target]].apply(pd.to_numeric, errors="coerce").dropna()
    # Avoid overplotting for large datasets.
    if len(plot_df) > 5000:
        plot_df = plot_df.sample(5000, random_state=RANDOM_STATE)
    fig, ax = plt.subplots()
    sns.regplot(data=plot_df, x=feature, y=target,
                scatter_kws={"alpha": 0.35, "s": 18},
                line_kws={"linewidth": 2}, ax=ax)
    ax.set_title(f"{name}: {feature} vs. {target}")
    plt.tight_layout()
    plt.show()
```

### 10. Identify the best model for each dataset
The lowest-RMSE model is selected for each task. Use the complete metric table and plots to discuss whether the best model also has a strong R² and sensible residual behavior.

```python
best_rows = (
    results_df.sort_values("RMSE")
    .groupby("Dataset", as_index=False)
    .first()
    .sort_values("Dataset")
)
display(best_rows[["Dataset", "Model", "MAE", "MSE", "RMSE", "R²"]].round(4))
best_rows.to_csv(OUTPUT_DIR / "best_model_per_dataset.csv", index=False)

for _, row in best_rows.iterrows():
    print(
        f"- {row['Dataset']}: best by test RMSE = {row['Model']} "
        f"(MAE={row['MAE']:.3f}, RMSE={row['RMSE']:.3f}, R²={row['R²']:.3f})."
    )
```

### 11. Inspect residual distributions
Residual distributions centered around zero suggest less systematic prediction bias; long tails can indicate outliers or difficult cases.

```python
for task in tasks:
    name = task["name"]
    best_model_name = best_rows.loc[best_rows["Dataset"] == name, "Model"].iloc[0]
    y_test, y_pred = predictions[name][best_model_name]
    residuals = np.asarray(y_test) - np.asarray(y_pred)
    fig, ax = plt.subplots()
    sns.histplot(residuals, kde=True, ax=ax)
    ax.axvline(0, linestyle="--")
    ax.set_title(f"{name}: residual distribution ({best_model_name})")
    ax.set_xlabel("Residual (actual − predicted)")
    plt.tight_layout()
    plt.show()
```
